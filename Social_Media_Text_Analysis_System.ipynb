```python
import nltk
import re
from nltk.sentiment import SentimentIntensityAnalyzer

nltk.download('vader_lexicon')

sia = SentimentIntensityAnalyzer()
```

```python
text = input("Enter a social media post: ")

# Sentiment analysis
score = sia.polarity_scores(text)

if score['compound'] >= 0.05:
    sentiment = "Positive 😊"
elif score['compound'] <= -0.05:
    sentiment = "Negative 😞"
else:
    sentiment = "Neutral 😐"

# Word count
words = re.findall(r'\b\w+\b', text)
word_count = len(words)

# Hashtags
hashtags = re.findall(r'#\w+', text)

# Mentions
mentions = re.findall(r'@\w+', text)

# Emojis / special symbols
emoji_count = len(re.findall(
    r'[😀-🙏🌀-🫿]', text
))

print("\n" + "="*45)
print("       SOCIAL MEDIA TEXT ANALYSIS")
print("="*45)

print("\nPost:")
print(text)

print("\n--- SENTIMENT ANALYSIS ---")
print("Positive Score :", round(score['pos'], 2))
print("Negative Score :", round(score['neg'], 2))
print("Neutral Score  :", round(score['neu'], 2))
print("Compound Score :", round(score['compound'], 2))
print("Final Sentiment:", sentiment)

print("\n--- TEXT ANALYSIS ---")
print("Word Count :", word_count)
print("Hashtags   :", hashtags if hashtags else "None")
print("Mentions   :", mentions if mentions else "None")
print("Emoji Count:", emoji_count)

print("\n--- CONCLUSION ---")

if sentiment.startswith("Positive"):
    print("👍 The post has a positive opinion.")
elif sentiment.startswith("Negative"):
    print("⚠️ The post has a negative opinion.")
else:
    print("ℹ️ The post has a neutral opinion.")

print("="*45)
```
